```python

import dlt
from pyspark.sql.functions import *

@dlt.table
def csv_python_2():
  return (
    spark.readStream.format("cloudFiles")
      .option("cloudFiles.format", "csv")
      .option("separator", ",")
      .option("header", "true")
      .load("/Volumes/my_catalog/my_schema/my_csv_python/")
  )
```
