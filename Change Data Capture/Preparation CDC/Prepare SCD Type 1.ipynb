```python
%sql
CREATE OR REPLACE streaming TABLE my_catalog.`1_bronze_db`.customer_bronze_raw2
  COMMENT "Raw data from customers CDC feed"
  TBLPROPERTIES(
    "quality" = "bronze",
    "pipelines.reset.allowed" = false
  )
  AS
SELECT 
  *, 
  CURRENT_TIMESTAMP() AS processing_time,
  _metadata.file_name AS source_file
FROM stream read_files(
  '/Volumes/my_catalog/my_schema/customers/customer1.json', -- Replace :source_vol with literal path
  format => "json",
  multiLine => 'true'
);
```
