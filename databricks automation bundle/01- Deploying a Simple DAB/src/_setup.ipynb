```sql
%python
class Config:
    def __init__(self):
        # Buscamos el widget 'catalog' que definió el YAML arriba
        try:
            self.catalog = dbutils.widgets.get("catalog")
        except:
            # Valor de respaldo por si lo corres a mano fuera del Job
            self.catalog = "dab_1_dev"
        
        self.bronze = f"{self.catalog}.bronze"
        self.silver = f"{self.catalog}.silver"
        self.gold   = f"{self.catalog}.gold"

    def create_schemas(self):
        spark.sql(f"CREATE SCHEMA IF NOT EXISTS {self.bronze}")
        spark.sql(f"CREATE SCHEMA IF NOT EXISTS {self.silver}")
        spark.sql(f"CREATE SCHEMA IF NOT EXISTS {self.gold}")

DA = Config()
DA.create_schemas()
```
